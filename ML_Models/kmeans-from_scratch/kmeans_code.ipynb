```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
```

## KMeans on 2D Data

```python
X,y = make_blobs(
    n_samples = 500,
    centers = 4,
    n_features = 2,
    cluster_std = 1.2,
    random_state = 42
)

plt.scatter(X[:,0], X[:,1], cmap = 'viridis')
plt.show()
```

```python
# Elbow method

from sklearn.cluster import KMeans

wcss = []
for i in range(1,11):
    km = KMeans(n_clusters = i)
    km.fit_predict(X)
    wcss.append(km.inertia_)
    
```

```python
wcss
plt.plot(range(1,11),wcss)
```

```python
km = KMeans(n_clusters= 4)
y_means = km.fit_predict(X)
y_means
```

```python
X[y_means == 3]
```

```python
plt.scatter(X[y_means == 0,0], X[y_means == 0, 1], color = 'yellow')
plt.scatter(X[y_means == 1,0], X[y_means == 1, 1], color = 'red')
plt.scatter(X[y_means == 2,0], X[y_means == 2, 1], color = 'blue')
plt.scatter(X[y_means == 3,0], X[y_means == 3, 1], color = 'black')
```

## KMeans on 3D Data

```python
centroids = [(-5,5,5),(5,5,-5),(3.5, -2.5, 4),(-2.5, 2.5, -4)]
cluster_std = [1,1,1,1]

X, y = make_blobs(n_samples= 200, cluster_std= cluster_std , centers= centroids, n_features= 1, random_state= 42)
```

```python
X
```

```python
import plotly.express as px
fig = px.scatter_3d(x = X[:,0],y = X[:,1], z = X[:,2])
fig.show()
```

```python
wcss = []
for i in range(1,11):
    km = KMeans(n_clusters= i)
    km.fit_predict(X)
    wcss.append(km.inertia_)
plt.plot(range(1,11),wcss)
```

```python
km = KMeans(n_clusters= 4)
y_pred = km.fit_predict(X)
df = pd.DataFrame()
df['col1'] = X[:,0]
df['col2'] = X[:,1]
df['col3'] = X[:,2]
labels = y_pred 
```

```python
fig = px.scatter_3d(df, x = 'col1',y = 'col2', z = 'col3', color = labels)
fig.show()
```
